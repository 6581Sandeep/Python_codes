# 2. ✅ Nutrition bar ingredient chart

Imagine you are a data analyst at a nutrition bar manufacturing company. Your department head approaches you with a question. The company produces a nutrition bar that contains **45g of raisins, 65g of almonds, and 30g of apricots**. The head of the manufacturing department wants you to create an ingredient percentage list for the nutrition bar using python.

Equipped with the knowledge of int, float and booleans in python, how would you approach this situation?

image.png

## this **1st way to solve**

```python
            # DONE BY SANDEEP SWAMI


# variables
raisins= 45
almonds= 65
apricots=30
total_weight_way_1 = raisins + almonds + apricots

# Calculate the percentage of raisins and print the variable

_raisins_percentage = (raisins / total_weight_way_1) * 100
print(_raisins_percentage)

# Calculate the percentage of almonds and print the variable

_almonds_percentage = (almonds / total_weight_way_1) * 100
print(_almonds_percentage)

# Calculate the percentage of apricots and print the variable

_apricots_percentage = (apricots / total_weight_way_1) * 100
print(_apricots_percentage)


```

## **2nd way to solve same problem for scalability**



```python
            # DONE BY SANDEEP SWAMI

# variables
ingredient = {
    "raisins" : 45,
    "almonds" : 65,
    "apricots" : 30
}

# we are doing sum all values
total_weight_way_2 = sum(_value for _name, _value in ingredient.items())

# we are storing the percentages to avoid repeat calculations
_ingredient_percentage = {
    _name +'_percent' : (_value / total_weight_way_2) * 100
    for _name, _value in ingredient.items()
}

# with help of this we can print the name and percent
for _name, _percent in _ingredient_percentage.items():
    print(f"{_name.capitalize()}: {_percent:.2f} %")
```

# 19. ✅📦📬🚚 Creating a shipping cost calculator! ⚙️

🌍 Domestic shipping within the country :

* Packages weighing up to 1 kg cost Rs.50 to ship 🤑
Each additional kg costs Rs.10 extra 💰

🌎 International shipping to other countries 🌏:

* Packages weighing up to 1 kg cost Rs.500 to ship 🤑

Each additional kg costs Rs.100 extra 💰
Ready to ship? 🚛🚀

```python

                    # Done by sandeep swami
def international(weight):
    base_price = 500  
    return base_price + (weight - 1) * 100 if weight > 1 else base_price

def domestic(weight):
    base_price = 50
    return base_price + (weight - 1) * 10 if weight > 1 else base_price


def details():
    try:
        shipping_type = int(input("""
    1. Type '1' for International. --> Each additional kg costs Rs.100 extra 💰
    2. Type '2' for Domestic. --> Each additional kg costs Rs.10 extra 💰
    
    Enter your shipping type (1 or 2): """))


        if shipping_type not in (1, 2):
            raise ValueError("Invalid shipping type! Please choose 1 or 2.")

        weight = int(input("\n    Enter weight in kg: "))

        if weight <= 0:
            raise ValueError("Weight must be greater than 0.")

        if shipping_type == 1:
            print(f"\n    Total cost: Rs.{international(weight)}")

        elif shipping_type == 2:
            print(f"\n    Total cost: Rs.{domestic(weight)}")

    except ValueError as e:
        print(f"\nError: {e}")
        print("Please enter valid numeric input.\n")
        details() 


details()
```

Example:

* Enter end number: 150

```python
# Write code here
def main():

    start = int(input("Enter start number: "))
    end = int(input("Enter end number: "))

    digit = int(input("Enter digit to count: "))
    count = 0

    for num in range(start, end + 1):
        temp = num

        while temp > 0:
            if temp % 10 == digit:
                count += 1
                
            temp //= 10

    print(f"The digit {digit} appears {count} times between {start} and {end}.")


main()

```

```python
# Define a simple calculator class

# Done by sandeep
class Calculator:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value

    def __sub__(self, other):
        return self.value - other.value

    def __mul__(self, other):
        return self.value * other.value

    def __truediv__(self, other):
        try:
            return self.value / other.value
        except ZeroDivisionError:
            return "Error: Division by zero!"


```

# 42. ✅ **Bank Account Management System** 💻🏦

Create a Bank Account Management System using Object-Oriented Programming (OOPs) in Python. The system should allow users to create and manage bank accounts, deposit and withdraw funds, and view account details.

bank.png

```python
# Define a BankAccount class


# Done by sandeep

class BankAccount:
    def __init__(self, name, account_number, balance=0):
        self.name = name.capitalize()              # public
        self._account_number = account_number  # protected
        self.__balance = balance            # private

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited.")
        else:
            print("Invalid amount.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid amount.")
        elif amount > self.__balance:
            print("Insufficient balance.")
        else:
            self.__balance -= amount
            print(f"₹{amount} withdrawn.")

    def get_balance(self):
        return self.__balance
    
    def show_details(self):
        print("\n--- Account Details ---")
        print(f"Name: {self.name}")
        print(f"Account Number: {self._account_number}")
        print(f"Balance: ₹{self.__balance}")
        print("----------------------")
        
```

```python
# Write your code here
class Time:
    def __init__(self, hours, minutes):
        self.hours = hours
        self.minutes = minutes

    def DisplayTime(self):
        if self.hours > 23:
            print("The input hours should be less than 24.")
            return
        if self.minutes > 59:
            print("The input minutes should be less than 60.")
            return

        # Determine AM or PM format
        if self.hours == 0:
            hour_display = 12
            period = "AM"
        elif self.hours == 12:
            hour_display = 12
            period = "PM"
        elif self.hours > 12:
            hour_display = self.hours - 12
            period = "PM"
        else:
            hour_display = self.hours
            period = "AM"

        print(f"The time is {hour_display}:{self.minutes:02d} {period}")

    def DisplayRatio(self):
        try:
            ratio = self.minutes / self.hours
            print(f"Ratio of minutes to hours is {ratio:.4f}...")
        except ZeroDivisionError:
            print("Cannot divide by zero hours!")


```
